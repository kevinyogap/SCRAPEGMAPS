```python
import pandas as pd
import json
import re

# Baca file JSON
with open('gmaps_wisata_dekat_danau_toba_1745854280.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

# Ubah ke DataFrame
df = pd.DataFrame(data)

# Fungsi untuk ekstrak koordinat dari link
def extract_coordinates(link):
    match = re.search(r'!3d(-?\d+\.\d+)!4d(-?\d+\.\d+)', link)
    if match:
        return float(match.group(1)), float(match.group(2))
    return None, None

# Tambahkan kolom latitude & longitude
df[['latitude', 'longitude']] = df['link'].apply(
    lambda x: pd.Series(extract_coordinates(x))
)

# Simpan ke Excel
df.to_excel('gmaps_danau_toba_with_coordinates.xlsx', index=False)

print("✅ Data disimpan ke 'gmaps_danau_toba_with_coordinates.xlsx' dengan koordinat.")
```
